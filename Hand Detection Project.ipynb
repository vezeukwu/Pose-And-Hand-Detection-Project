```python
import cv2
import mediapipe as mp
```

```python
# Identifying the webcam
capture = cv2.VideoCapture(0)
```

```python
# Leveraging the mediapipe library for hand detection

mpHands = mp.solutions.hands
hands = mpHands.Hands()
mpDraw = mp.solutions.drawing_utils
```

```python
# Switching on the webcam

while True:
    _, img = capture.read()

# Convert image from BGR to RGB
    imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Applying media pipe
    results = hands.process(imgRGB)

   # print(results.multi_hand_landmarks)

    if results.multi_hand_landmarks:
        for handlms in results.multi_hand_landmarks:
            for id, lm in enumerate(handlms.landmark):
               # print(id, lm)

                mpDraw.draw_landmarks(img, handlms, mpHands.HAND_CONNECTIONS)

    cv2.putText(img, "Hand Detection", (10,40), cv2.FONT_HERSHEY_PLAIN, 2, (100,255,200), 2)
    cv2.imshow("Hand Detection", img)
    if cv2.waitKey(0) & 0xff == ord("q"):
        break
# Release the capture when all the process is concluded
capture.release()
cv2.destroyAllWindows()
```
